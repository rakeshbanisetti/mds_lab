```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the new dataset: advertising.csv
df = pd.read_csv('/content/advertising.csv')

# Display the first 5 rows to verify the data loading and see column names
print("Dataset loaded successfully. First 5 rows:")
display(df.head())

# Display column information to identify suitable attributes for plotting
print("\nColumn Information:")
df.info()
```

### Data Visualization

Now we will generate various plots. **Please replace `column_1` and `column_2` with two actual column names from your dataset** that are suitable for the respective plot types. For example, for a scatter plot, both should be numerical. For a bar plot, one might be categorical and the other numerical.

```python
# --- 1. Scatter Plot (Relationship between two numerical variables) ---
# Using 'TV' and 'Sales' to visualize the relationship between TV advertising spend and sales.

plt.figure(figsize=(10, 6))
sns.scatterplot(x=df['TV'], y=df['Sales'])
plt.title('Scatter Plot of Sales vs. TV Advertising Spend')
plt.xlabel('TV Advertising Spend')
plt.ylabel('Sales')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# --- 2. Line Plot (Trends) ---
# As there's no explicit time-series column in this dataset, we will plot Sales against the DataFrame index.
# This can show a general trend in the order the data appears, but is not a time-based trend.

plt.figure(figsize=(12, 6))
sns.lineplot(x=df.index, y=df['Sales'])
plt.title('Line Plot of Sales (by Data Entry Order)')
plt.xlabel('Data Entry Order (Index)')
plt.ylabel('Sales')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# --- 3. Bar Plot (Aggregation of numerical data by categories) ---
# Creating bins for 'TV' advertising spend and plotting the average 'Sales' for each bin.

# Create categories for 'TV' spend
df['TV_bins'] = pd.cut(df['TV'], bins=3, labels=['Low TV', 'Medium TV', 'High TV'])

# Calculate mean sales for each TV bin
df_agg = df.groupby('TV_bins')['Sales'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x=df_agg['TV_bins'], y=df_agg['Sales'], palette='viridis')
plt.title('Average Sales by TV Advertising Spend Category')
plt.xlabel('TV Advertising Spend Category')
plt.ylabel('Average Sales')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# --- 4. Histogram (Distribution of a single numerical variable) ---
# Using 'Sales' to show its distribution.

plt.figure(figsize=(10, 6))
sns.histplot(df['Sales'], kde=True, bins=15) # kde=True adds a density curve
plt.title('Histogram of Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# --- 5. Density Plot (Smoothed distribution of a single numerical variable) ---
# Using 'Sales' to show its smoothed distribution.

plt.figure(figsize=(10, 6))
sns.kdeplot(df['Sales'], fill=True, color='purple')
plt.title('Density Plot of Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Density')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Visualizations for `house_prices.csv` (New Cells)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the house_prices.csv dataset into a new DataFrame
df_house_prices = pd.read_csv('/content/house_prices.csv')

# Display the first 5 rows to verify the data loading and see column names
print("House Prices Dataset loaded successfully. First 5 rows:")
display(df_house_prices.head())

# Display column information to identify suitable attributes for plotting
print("\nHouse Prices Column Information:")
df_house_prices.info()
```

```python
# --- 1. Scatter Plot (House Prices: sqft_living vs. price) ---
# Using 'sqft_living' and 'price' from df_house_prices.

plt.figure(figsize=(10, 6))
sns.scatterplot(x=df_house_prices['sqft_living'], y=df_house_prices['price'])
plt.title('Scatter Plot of House Price vs. Living Area (House Prices Dataset)')
plt.xlabel('Square Footage of Living Area')
plt.ylabel('Price')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# --- 2. Line Plot (House Prices: Price over Time) ---
# Converting 'date' to datetime and plotting price trends over time from df_house_prices.

# Convert 'date' column to datetime objects
df_house_prices['date'] = pd.to_datetime(df_house_prices['date'])

# Ensure data is sorted by date for a meaningful line plot
df_house_prices_sorted = df_house_prices.sort_values(by='date')

plt.figure(figsize=(14, 7))
sns.lineplot(x=df_house_prices_sorted['date'], y=df_house_prices_sorted['price'])
plt.title('Line Plot of House Price over Time (House Prices Dataset)')
plt.xlabel('Date')
plt.ylabel('Price')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# --- 3. Bar Plot (House Prices: Count by Bedrooms) ---
# Using 'bedrooms' to show the count of houses by number of bedrooms from df_house_prices.

plt.figure(figsize=(10, 6))
sns.countplot(x=df_house_prices['bedrooms'], palette='viridis')
plt.title('Bar Plot of House Counts by Number of Bedrooms (House Prices Dataset)')
plt.xlabel('Number of Bedrooms')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# --- 4. Histogram (House Prices: Price Distribution) ---
# Using 'price' to show its distribution from df_house_prices.

plt.figure(figsize=(10, 6))
sns.histplot(df_house_prices['price'], kde=True, bins=50) # kde=True adds a density curve
plt.title('Histogram of House Price Distribution (House Prices Dataset)')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# --- 5. Density Plot (House Prices: Price Distribution) ---
# Using 'price' to show its smoothed distribution from df_house_prices.

plt.figure(figsize=(10, 6))
sns.kdeplot(df_house_prices['price'], fill=True, color='purple')
plt.title('Density Plot of House Price Distribution (House Prices Dataset)')
plt.xlabel('Price')
plt.ylabel('Density')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
