```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {
    'Student': [f'S{i}' for i in range(1, 16)],
    'Score': [65, 72, 88, 55, 91, 78, 62, 85, 70, 95, 60, 75, 80, 50, 99]
}
df_scores = pd.DataFrame(data)

print("Original DataFrame with Scores:")
display(df_scores)

# Apply pd.cut() to create 3 equal-width bins
df_scores['Score_Category_Cut'] = pd.cut(
    df_scores['Score'],
    bins=3,
    labels=['Low', 'Medium', 'High'],
    include_lowest=True
)

print("\nDataFrame after applying pd.cut() (Equal-Width Bins):")
display(df_scores.head())

print("\nValue counts for Score_Category_Cut:")
display(df_scores['Score_Category_Cut'].value_counts().sort_index())
```

**`pd.qcut()` (Equal-Depth Bins / Quantile-Based Bins)**

*   **Purpose**: `pd.qcut()` is used to segment data into bins such that **each bin has roughly the same number of observations (or the same quantile range)**. It determines bin edges based on the quantiles of the data.
*   **Use Case**: Ideal when you want to create groups with an approximately equal number of members, regardless of the numerical range they cover. For example, dividing students into 'Bottom 25%', 'Middle 50%', 'Top 25%' based on their scores.
*   **Disadvantage**: The bin widths can vary significantly, especially with skewed data, which might make the bin ranges less intuitive or interpretable compared to `pd.cut()`.

**Example for `pd.qcut()`:** Now, let's use `pd.qcut()` on the same 'Score' column to create three bins, each containing approximately the same number of students.

```python
import pandas as pd
import numpy as np

# Using the same df_scores DataFrame from the previous example
# Apply pd.qcut() to create 3 equal-depth bins
df_scores['Score_Category_Qcut'] = pd.qcut(
    df_scores['Score'],
    q=3,
    labels=['Bottom', 'Middle', 'Top'],
    duplicates='drop' # Handle cases where quantile edges are not unique
)

print("DataFrame after applying pd.qcut() (Equal-Depth Bins):")
display(df_scores.head())

print("\nValue counts for Score_Category_Qcut:")
display(df_scores['Score_Category_Qcut'].value_counts().sort_index())
```

### Key Differences Summarized:

*   **`pd.cut()`**: Creates bins of **equal *width* (range)**. The number of data points in each bin can vary greatly.
*   **`pd.qcut()`**: Creates bins of **equal *depth* (number of observations)**. The width of each bin can vary greatly.

Choose `pd.cut()` when you have specific, meaningful ranges you want to define. Choose `pd.qcut()` when you want to ensure an even distribution of data points across your bins, regardless of the numerical range they span.

### Re-loading and Cleaning Data for `loc` and `iloc` Demonstrations

Given that the Excel file `/content/research_student (1).xlsx` is now available, I will re-load the data and perform the cleaning steps to ensure `research_student_df_cleaned` is correctly defined for the following `loc` and `iloc` operations.

```python
import pandas as pd

excel_file_path = '/content/research_student (1).xlsx'

try:
    # Load the Excel file into a pandas DataFrame
    research_student_df = pd.read_excel(excel_file_path)
    print(f"Successfully loaded '{excel_file_path}' into a DataFrame with {research_student_df.shape[0]} rows and {research_student_df.shape[1]} columns.")

    # Display the first few rows to confirm
    print("\nFirst 5 rows of the 'research_student_df' DataFrame:")
    display(research_student_df.head())

except Exception as e:
    print(f"An error occurred while loading the Excel file: {e}")
```

```python
# Drop rows with any missing values to create the cleaned DataFrame
research_student_df_cleaned = research_student_df.dropna()

print(f"Shape of DataFrame before dropping missing values: {research_student_df.shape}")
print(f"Shape of DataFrame after dropping missing values: {research_student_df_cleaned.shape}")

print("\nFirst 5 rows of the cleaned 'research_student_df_cleaned' DataFrame:")
display(research_student_df_cleaned.head())
```

### Demonstrating `loc` operations

`loc` is primarily label-based, meaning you use the actual row and column labels to select data. It can also take a boolean array.

```python
# Select a single row by its index label (e.g., the first row, which has index 1 after cleaning)
print("\n--- Using loc to select a single row (index 1) ---")
display(research_student_df_cleaned.loc[1])

# Select multiple rows by their index labels and specific columns
print("\n--- Using loc to select rows 2, 3, and columns 'Branch', 'Marks[10th]' ---")
display(research_student_df_cleaned.loc[[2, 3], ['Branch', 'Marks[10th]']])

# Select a slice of rows and all columns
print("\n--- Using loc to select a slice of rows (from index 5 to 7, inclusive) and all columns ---")
display(research_student_df_cleaned.loc[5:7, :])

# Select rows based on a boolean condition and specific columns
print("\n--- Using loc with a boolean condition (Marks[10th] > 90) and columns 'Branch', 'Marks[10th]', 'Marks[12th]' ---")
display(research_student_df_cleaned.loc[research_student_df_cleaned['Marks[10th]'] > 90, ['Branch', 'Marks[10th]', 'Marks[12th]']].head())
```

### Demonstrating `iloc` operations

`iloc` is primarily integer-position-based, meaning you use the numerical position (from 0 to N-1) of rows and columns to select data.

```python
# Select a single row by its integer position (e.g., the first row, at position 0)
print("\n--- Using iloc to select a single row (position 0) ---")
display(research_student_df_cleaned.iloc[0])

# Select multiple rows by their integer positions and specific columns by integer positions
print("\n--- Using iloc to select rows at positions 1, 2 and columns at positions 0 (Branch), 1 (Marks[10th]) ---")
display(research_student_df_cleaned.iloc[[1, 2], [0, 1]])

# Select a slice of rows and all columns
print("\n--- Using iloc to select a slice of rows (from position 4 up to, but not including, 7) and all columns ---")
display(research_student_df_cleaned.iloc[4:7, :])

# Select all rows and a slice of columns (e.g., the first 3 columns)
print("\n--- Using iloc to select all rows and a slice of columns (first 3 columns) ---")
display(research_student_df_cleaned.iloc[:, 0:3].head())
```
