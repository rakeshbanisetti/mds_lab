### Comparing Basic Operations: Python Lists vs. NumPy Arrays

Now, let's compare the time taken for common arithmetic operations (addition, subtraction, multiplication, and division) using both standard Python lists and NumPy arrays. We will use a large number of elements to clearly see the performance differences.

```python
import time
import numpy as np

operation_size = 1000000 # Using the same size for consistency

# --- Python List Operations ---
python_list_1 = list(range(operation_size))
python_list_2 = list(range(operation_size, operation_size * 2))

print(f"\n--- Python List Operations (on {operation_size} elements) ---")

# List Addition
start_time = time.time()
python_list_add = [python_list_1[i] + python_list_2[i] for i in range(operation_size)]
end_time = time.time()
time_taken_list_add = end_time - start_time
print(f"Time for List Addition: {time_taken_list_add:.6f} seconds")

# List Subtraction
start_time = time.time()
python_list_sub = [python_list_1[i] - python_list_2[i] for i in range(operation_size)]
end_time = time.time()
time_taken_list_sub = end_time - start_time
print(f"Time for List Subtraction: {time_taken_list_sub:.6f} seconds")

# List Multiplication
start_time = time.time()
python_list_mul = [python_list_1[i] * python_list_2[i] for i in range(operation_size)]
end_time = time.time()
time_taken_list_mul = end_time - start_time
print(f"Time for List Multiplication: {time_taken_list_mul:.6f} seconds")

# List Division
# Avoid ZeroDivisionError for division, add 1 to list_2 to ensure non-zero denominators
python_list_2_div = list(range(operation_size, operation_size * 2))
start_time = time.time()
python_list_div = [python_list_1[i] / (python_list_2_div[i] if python_list_2_div[i] != 0 else 1) for i in range(operation_size)]
end_time = time.time()
time_taken_list_div = end_time - start_time
print(f"Time for List Division: {time_taken_list_div:.6f} seconds")
```

```python
import time
import numpy as np

operation_size = 1000000 # Using the same size for consistency

# --- NumPy Array Operations ---
numpy_array_1 = np.arange(operation_size)
numpy_array_2 = np.arange(operation_size, operation_size * 2)

print(f"\n--- NumPy Array Operations (on {operation_size} elements) ---")

# NumPy Array Addition
start_time = time.time()
numpy_array_add = numpy_array_1 + numpy_array_2
end_time = time.time()
time_taken_np_add = end_time - start_time
print(f"Time for NumPy Array Addition: {time_taken_np_add:.6f} seconds")

# NumPy Array Subtraction
start_time = time.time()
numpy_array_sub = numpy_array_1 - numpy_array_2
end_time = time.time()
time_taken_np_sub = end_time - start_time
print(f"Time for NumPy Array Subtraction: {time_taken_np_sub:.6f} seconds")

# NumPy Array Multiplication
start_time = time.time()
numpy_array_mul = numpy_array_1 * numpy_array_2
end_time = time.time()
time_taken_np_mul = end_time - start_time
print(f"Time for NumPy Array Multiplication: {time_taken_np_mul:.6f} seconds")

# NumPy Array Division
# Avoid ZeroDivisionError, ensure non-zero denominators
numpy_array_2_div = np.arange(operation_size, operation_size * 2)
# Replace zeros in array_2_div with a non-zero value if any, though arange won't have zeros here
# For robustness, we could add a small epsilon or check for zeros if the data source was different.
start_time = time.time()
numpy_array_div = numpy_array_1 / numpy_array_2_div
end_time = time.time()
time_taken_np_div = end_time - start_time
print(f"Time for NumPy Array Division: {time_taken_np_div:.6f} seconds")
```

The output of the previous cells demonstrates the significant performance improvement that NumPy arrays offer over standard Python lists for element-wise arithmetic operations, especially with a large number of elements. This is due to NumPy's optimized C implementations for these operations.

### Using `%%timeit` for Micro-benchmarking

The `%%timeit` magic command (available in Jupyter and Colab notebooks) is used to measure the execution time of a single statement or an entire cell. It automatically runs the code multiple times (typically 100,000 to 10,000,000 times for fast code) and calculates the average execution time, providing a more reliable measurement for small operations.

Let's use `%%timeit` to compare the basic operations on our small lists `l1` and `l2` and their NumPy array equivalents.

```python
import numpy as np

l1 = [55,2,3,4,5]
l2 = [55,7,8,9,10]

numpy_arr_l1 = np.array(l1)
numpy_arr_l2 = np.array(l2)
```

### Creating a Matrix with All Same Elements

```python
import numpy as np

matrix_full = np.full((3, 3), 7)
matrix_full[0,0] = 55
print("Matrix using np.full():")
print(matrix_full)

matrix_zeros_multiplied = np.zeros((2, 4)) * 5
matrix_zeros_multiplied[0,0] = 55
print("\nMatrix using np.zeros() and multiplication:")
print(matrix_zeros_multiplied)

matrix_ones_multiplied = np.ones((4, 2)) * 10
matrix_ones_multiplied[0,0] = 55
print("\nMatrix using np.ones() and multiplication:")
print(matrix_ones_multiplied)
```

### Creating an Identity Matrix

An identity matrix is a square matrix where all the elements on the main diagonal are ones, and all other elements are zeros. It is denoted by $I_n$, where $n$ is the dimension of the square matrix. In NumPy, you can create an identity matrix using the `np.eye()` function.

```python
import numpy as np

identity_matrix_3x3 = np.eye(3)
identity_matrix_3x3[0,0] = 55
print("3x3 Identity Matrix:")
print(identity_matrix_3x3)

identity_matrix_4x4 = np.eye(4)
identity_matrix_4x4[0,0] = 55
print("\n4x4 Identity Matrix:")
print(identity_matrix_4x4)

identity_matrix_float = np.eye(2, dtype=float)
identity_matrix_float[0,0] = 55
print("\n2x2 Identity Matrix with float type:")
print(identity_matrix_float)
```

### Creating a Diagonal Matrix with Specific Values

To create a matrix where the diagonal elements are a specific value (e.g., 5) and all other elements are zero, you can simply multiply an identity matrix by that scalar value. An identity matrix has ones on the diagonal, so scaling it will place the desired value on the diagonal.

```python
import numpy as np
diagonal_matrix_5 = np.eye(3) * 5
diagonal_matrix_5[0,0] = 55
print("3x3 Matrix with 5s on the diagonal:")
print(diagonal_matrix_5)
diagonal_matrix_10 = np.eye(4) * 10
diagonal_matrix_10[0,0] = 55
print("\n4x4 Matrix with 10s on the diagonal:")
print(diagonal_matrix_10)
```
